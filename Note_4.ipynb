```python
import numpy as np
import time

# -------------------------------
# Define System Unit Variations
# -------------------------------
system_units = {
    "Desktop_PC": {"CPU_cores": 8, "CPU_speed_GHz": 3.6, "RAM_GB": 32, "Storage_GB": 1024, "GPU": True},
    "All_in_One": {"CPU_cores": 6, "CPU_speed_GHz": 3.2, "RAM_GB": 16, "Storage_GB": 512, "GPU": True},
    "Notebook": {"CPU_cores": 4, "CPU_speed_GHz": 2.8, "RAM_GB": 16, "Storage_GB": 512, "GPU": False},
    "Tablet": {"CPU_cores": 2, "CPU_speed_GHz": 2.0, "RAM_GB": 8, "Storage_GB": 256, "GPU": False}
}

# -------------------------------
# CPU Simulation Function
# -------------------------------
def simulate_cpu_workload(n_tasks, CPU_cores, CPU_speed_GHz):
    start = time.time()
    tasks_per_core = n_tasks // CPU_cores
    for _ in range(CPU_cores):
        _ = np.sum(np.arange(tasks_per_core) ** 2)  # heavy computation
    end = time.time()
    # Adjust for CPU speed
    return end - start / CPU_speed_GHz

# -------------------------------
# RAM Simulation Function
# -------------------------------
def simulate_ram_usage(data_size_GB, RAM_GB):
    if data_size_GB > RAM_GB:
        return f"Not enough RAM! Dataset: {data_size_GB}GB, RAM: {RAM_GB}GB"
    else:
        return f"RAM sufficient: Dataset {data_size_GB}GB fits in {RAM_GB}GB RAM"

# -------------------------------
# Simulate Workload for Each System Unit
# -------------------------------
tasks = 10_000_000  # Monte Carlo tasks for financial simulation
dataset_GB = 10     # Example large dataset for biostatistics

for name, config in system_units.items():
    cpu_time = simulate_cpu_workload(tasks, config["CPU_cores"], config["CPU_speed_GHz"])
    ram_status = simulate_ram_usage(dataset_GB, config["RAM_GB"])
    print(f"\nSystem Unit: {name}")
    print(f"CPU simulation time: {cpu_time:.2f} seconds")
    print(f"RAM check: {ram_status}")
    print(f"Configuration: {config}")
```

# Processor Components: Control Unit, ALU, Machine Cycle, and System Clock

---

## 1. The Control Unit (CU)
- **Function:** Directs and coordinates the operations of the computer.  
- Acts like a **traffic cop**, interpreting program instructions and initiating the appropriate actions.  
- Directs other components, including:  
  - **Arithmetic Logic Unit (ALU)**  
  - **Buses** that carry data between CPU, memory, and peripherals

---

## 2. The Arithmetic Logic Unit (ALU)
- Performs **arithmetic, comparison, and logic operations**  
- **Arithmetic operations:** addition, subtraction, multiplication, division  
- **Comparison operations:** Determines whether one data item is greater than, equal to, or less than another, guiding subsequent actions  

---

## 3. Machine Cycle
Every instruction processed by the CPU follows a **machine cycle** with four steps:  

1. **Fetching:** Retrieve instruction or data from memory  
2. **Decoding:** Translate instruction into signals the CPU can execute  
3. **Executing:** Perform the requested operation (e.g., calculation or comparison)  
4. **Storing:** Write results back to memory (not permanent storage)  

**Example:** Calculating 100 × 48  
- **Step 1:** Control Unit fetches instruction and data from memory  
- **Step 2:** Control Unit decodes instruction and sends data to ALU  
- **Step 3:** ALU performs the calculation → 100 × 48 = 4800  
- **Step 4:** Result is stored in memory and displayed if needed  

> *(Figure 5 illustrates this machine cycle visually.)*

---



image.png


## 4. The System Clock
- Small quartz crystal circuit controlling **timing of all CPU operations**  
- Produces **regular electronic pulses (“ticks”)** that pace instruction execution  
- **Clock speed** measured in hertz (cycles per second)  
  - **1 GHz = 1 billion ticks per second**  
  - Example: a 3 GHz CPU executes **3 billion cycles per second**  

**Performance depends on:**  
- Processor type and architecture  
- Cache size  
- Memory access times  
- Bus width and bus clock speed  

> Faster clock speeds allow the CPU to execute more instructions per second, but overall performance depends on the CPU architecture and supporting components.

---

## 5. Practical Context
- **Financial Engineers:**  
  - Fast clock and multi-core processors accelerate **portfolio simulations**, **risk modeling**, and **algorithmic trading**, where billions of calculations may occur per second  

- **Biostatisticians:**  
  - High clock speeds and efficient ALU operation speed up **data-intensive modeling**, such as **genome-wide association studies** or **large-scale epidemiological simulations**

```python
import time
import numpy as np

# -------------------------------
# Define Processor Parameters
# -------------------------------
cpu = {
    "cores": 4,
    "clock_GHz": 3.0,     # System clock
    "cache_MB": 8,
}

# -------------------------------
# ALU Function: Arithmetic + Comparison
# -------------------------------
def ALU(operation, a, b):
    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b
    elif operation == "multiply":
        return a * b
    elif operation == "divide":
        return a / b
    elif operation == "compare":
        if a > b:
            return "greater"
        elif a < b:
            return "less"
        else:
            return "equal"

# -------------------------------
# Machine Cycle Simulation
# -------------------------------
def machine_cycle(instructions):
    results = []
    for instr in instructions:
        # Step 1: Fetch (simulate delay)
        time.sleep(0.00001)  # simulate memory access

        # Step 2: Decode
        op, a, b = instr

        # Step 3: Execute
        res = ALU(op, a, b)

        # Step 4: Store (here we just save in results)
        results.append(res)
    return results

# -------------------------------
# Generate Instructions (Example)
# -------------------------------
# Financial: simulate 1 million simple calculations
num_tasks = 100_000
np.random.seed(42)
instructions = [("multiply", np.random.rand(), np.random.rand()) for _ in range(num_tasks)]

# -------------------------------
# Simulate Execution with System Clock
# -------------------------------
start_time = time.time()
results = machine_cycle(instructions)
end_time = time.time()

# Adjust execution time by CPU clock speed (higher GHz → faster)
adjusted_time = (end_time - start_time) / cpu["clock_GHz"]

print(f"Executed {num_tasks} ALU instructions")
print(f"Raw execution time: {end_time - start_time:.2f} s")
print(f"Adjusted by {cpu['clock_GHz']} GHz CPU clock: {adjusted_time:.2f} s")

```

## 5. From Keyboard to Screen
When a user types a character, the computer performs the following steps:  

1. **Input:** User presses a key (e.g., capital `T`). Keyboard generates a **scan code**  
2. **Transmission:** Scan code is sent to the system unit  
3. **Conversion:** System unit converts the scan code into **binary form (ASCII)** and stores it in memory  
4. **Processing & Output:** CPU processes the binary data, and software converts it back into a **human-readable character** on the screen or printout

```python
# -------------------------------
# Data Representation Simulation
# -------------------------------
import numpy as np

# Example 1: Convert characters to binary (ASCII)
def char_to_binary(char):
    """Convert a character to 8-bit binary using ASCII"""
    return format(ord(char), '08b')

# Example 2: Convert numbers to binary
def number_to_binary(number):
    """Convert integer 0-255 to 8-bit binary"""
    if 0 <= number <= 255:
        return format(number, '08b')
    else:
        raise ValueError("Number must be between 0 and 255")

# Example 3: Simulate a string as bytes
def string_to_bytes(string):
    """Convert a string to a list of ASCII byte values"""
    return [char_to_binary(c) for c in string]

# -------------------------------
# Practical Examples
# -------------------------------
# Character to binary
char = 'E'
binary_char = char_to_binary(char)
print(f"Character '{char}' in binary: {binary_char}")

# Number to binary
num = 6
binary_num = number_to_binary(num)
print(f"Number {num} in binary: {binary_num}")

# String to binary bytes
text = "TRADE"
binary_text = string_to_bytes(text)
print(f"String '{text}' as binary bytes:")
for c, b in zip(text, binary_text):
    print(f"'{c}' → {b}")

# -------------------------------
# Simulate Binary Data for Large Datasets
# -------------------------------
# Financial context: encode 1 million transaction IDs (0-255)
transaction_ids = np.random.randint(0, 256, size=10**6)
binary_transactions = [number_to_binary(x) for x in transaction_ids[:5]]  # show first 5
print("\nSample binary transactions (first 5):", binary_transactions)

# Biostatistics context: encode genomic ACGT sequences as binary
genome_seq = "ACGTACGT"
# Simple encoding: A=00, C=01, G=10, T=11
binary_genome = ''.join({'A':'00','C':'01','G':'10','T':'11'}[nuc] for nuc in genome_seq)
print("Genome sequence:", genome_seq)
print("Binary representation:", binary_genome)
```

# Memory

---

## 1. Definition and Role
- Memory consists of **electronic components** that store instructions, data, and processing results.  
- Located on the **motherboard** or other circuit boards.  
- **Stores three primary categories:**  
  1. **System Software:** Operating system and programs that maintain the computer  
  2. **Application Programs:** Software for specific tasks (e.g., financial modeling, statistical analysis)  
  3. **Data:** Information being processed by applications  

> This dual role of storing both programs and data is called the **stored-program concept**.

---

## 2. Bytes and Addressable Memory
- **Byte:** Basic storage unit in memory (8 bits)  
- Each byte has a **unique memory address**, similar to seats in an opera house:  
  - A seat holds one person; a memory location holds one byte  
  - Both seats and memory locations can be empty  
- **Accessing memory:** Referencing the address of a byte to **read or write data**

---

## 3. Memory Sizes

| Term       | Abbreviation | Approx. Bytes | Exact Bytes      |
|------------|--------------|---------------|----------------|
| Kilobyte   | KB           | 1,000         | 1,024          |
| Megabyte   | MB           | 1 million     | 1,048,576      |
| Gigabyte   | GB           | 1 billion     | 1,073,741,824  |
| Terabyte   | TB           | 1 trillion    | 1,099,511,627,776 |

- **RAM requirements:**  
  - Basic home applications: 1 GB  
  - Business applications (spreadsheets, accounting): 2–8 GB  
  - Graphics-intensive or professional applications: 8 GB+

---

## 4. Types of Memory

### Volatile vs Nonvolatile
- **Volatile Memory:** Loses contents when power is off (e.g., RAM)  
- **Nonvolatile Memory:** Retains contents without power (e.g., ROM, Flash, CMOS)

### RAM (Random Access Memory)
- Can be **read and written by the CPU**  
- Programs and data load into RAM during execution  

**Types of RAM:**  
- **Dynamic RAM (DRAM):** Requires constant refreshing; widely used  
- **Static RAM (SRAM):** Faster, more reliable; less frequent refreshing  
- **Magnetoresistive RAM (MRAM):** Uses magnetic charges; fast and energy-efficient

### Cache
- Stores frequently used instructions and data for **faster access**  
- **L1 Cache:** Small, on the processor chip (8 KB–128 KB)  
- **L2 Cache:** Larger, slightly slower (64 KB–16 MB)  
- Advanced processors may use **transfer cache** for even faster performance

### ROM (Read-Only Memory)
- Stores **permanent data and instructions**; usually not modifiable  
- **Firmware:** ROM written at manufacture (e.g., system startup instructions)

### Flash Memory
- **Nonvolatile** and **electronically rewritable**  
- Used in mobile devices, portable media players, and computers

### CMOS (Complementary Metal-Oxide Semiconductor)
- High-speed, low-power memory  
- Often **battery-backed** to retain information like date, time, and startup data even when the computer is off

---

## 5. Memory Access Time
- **Access time:** Time for the processor to read/write data from memory  
- Memory is much faster than storage devices like hard disks  
- Measured in fractions of a second:  
  - Millisecond (ms) = 1/1,000 sec  
  - Microsecond (µs) = 1/1,000,000 sec  
  - Nanosecond (ns) = 1/1,000,000,000 sec  
  - Picosecond (ps) = 1/1,000,000,000,000 sec  

> Faster memory access improves overall **computer performance**

---

## 6. Practical Context
- **Financial Engineers:**  
  - Large simulations and **real-time trading** rely on fast RAM and cache  
  - Multi-gigabyte datasets in memory accelerate **portfolio and risk calculations**  

- **Biostatisticians:**  
  - Genomic and clinical datasets load into RAM for **statistical modeling, analysis, and visualization**  
  - Cache memory reduces **latency** during repeated calculations


```python
import numpy as np
import time

# -------------------------------
# Define Memory Components
# -------------------------------
memory_system = {
    "L1_cache": {"size_KB": 64, "access_ns": 1},       # L1 cache
    "L2_cache": {"size_KB": 512, "access_ns": 4},     # L2 cache
    "RAM": {"size_GB": 16, "access_ns": 100},         # Main memory
    "Flash": {"size_GB": 512, "access_us": 100}       # Secondary storage
}

# -------------------------------
# Simulate Memory Access
# -------------------------------
def access_memory(data_size_bytes, memory_type):
    """Simulate access time based on memory type"""
    mem = memory_system[memory_type]

    # Convert all access times to seconds
    if "access_ns" in mem:
        access_time = mem["access_ns"] * 1e-9
    elif "access_us" in mem:
        access_time = mem["access_us"] * 1e-6
    else:
        access_time = 1e-3  # default 1 ms

    # Simulate accessing each byte
    # For large datasets, we sample to avoid huge loops
    sample_size = min(data_size_bytes, 10**6)
    start = time.time()
    _ = np.sum(np.random.randint(0, 256, sample_size))  # dummy computation
    end = time.time()

    # Effective access time includes memory latency
    effective_time = (end - start) + sample_size * access_time
    return effective_time

# -------------------------------
# Practical Simulation
# -------------------------------
# Example datasets
financial_dataset_bytes = 500 * 1024 * 1024      # 500 MB
genomic_dataset_bytes = 2 * 1024 * 1024 * 1024   # 2 GB

# Simulate access for Financial Engineer
fe_L1_time = access_memory(financial_dataset_bytes, "L1_cache")
fe_RAM_time = access_memory(financial_dataset_bytes, "RAM")

# Simulate access for Biostatistician
bio_L2_time = access_memory(genomic_dataset_bytes, "L2_cache")
bio_RAM_time = access_memory(genomic_dataset_bytes, "RAM")

print("Financial Engineer - Access Times:")
print(f"L1 Cache: {fe_L1_time:.4f} s")
print(f"RAM: {fe_RAM_time:.4f} s")

print("\nBiostatistician - Access Times:")
print(f"L2 Cache: {bio_L2_time:.4f} s")
print(f"RAM: {bio_RAM_time:.4f} s")
```

#A4 Expansion Slots and Adapter Cards

---

## 1. Expansion Slots
- An **expansion slot** is a socket on the motherboard designed to hold an **adapter card** (also called an expansion card).  
- Expansion slots allow the system to **enhance its functionality** or **connect to external devices** (peripherals).

---

## 2. Adapter Cards
- **Adapter cards** are circuit boards that **extend or improve** the functions of the system unit.  
- **Peripherals** are devices controlled by the computer’s processor (e.g., modems, printers, scanners, keyboards, disk drives).  

**Examples of Adapter Cards:**

| Adapter Card | Purpose |
|--------------|---------|
| Sound        | Enhances sound output; allows input from microphones and output to speakers or headphones |
| Video/Graphics | Converts computer output into a video signal for the monitor |
| Network      | Connects to other computers and peripherals |
| Modem        | Connects to the internet via telephone or cable lines |
| Disk Controller | Connects and controls disk drives |
| USB          | Connects USB devices (flash drives, peripherals) |
| FireWire     | Connects FireWire devices |
| TV Tuner / CableCARD / HDTV tuner | Allows viewing of television or cable broadcasts on the computer monitor |
| PC-to-TV Converter | Connects a computer to a television |
| MIDI         | Connects musical instruments |
| Video Capture | Connects analog video cameras or VCRs |

> Some motherboards **integrate sound and video functionality**, while others require separate adapter cards.  
> Example: a **sound card** allows audio input/output, while a **video card** generates video signals for the display.

---

## 3. Removable Flash Memory
- Provides **portable, rewritable storage** for computers and devices.  

**Common Types:**
- **Memory Cards:** Small, removable devices used in cameras, smartphones, portable media players, and card readers  
- **USB Flash Drives:** Plug into USB ports for portable storage of data and programs  
- **PC Cards / ExpressCard Modules:** Thin, credit-card-sized expansion cards for notebooks and Tablet PCs  
  - Provide memory expansion, wireless access, or other functionalities  
  - ExpressCard modules are **smaller than traditional PC Cards**, often about half the size  

> Many modern devices, such as smartphones, tablets, and portable media players, rely on flash memory for **storage and software updates**.  
> Flash memory can be **easily removed, transferred, and reused**, making it ideal for **data mobility and backup**.

---

## 4. Practical Context

  - Adapter cards and high-speed removable memory enable **low-latency data acquisition** and **real-time trading simulations**  
  - Video and network cards support **multi-monitor setups** and **high-throughput connectivity**  

- **Biostatisticians:**  
  - Removable flash memory allows transfer of **large genomic datasets**, **clinical trial data**, or **research results** between workstations and servers  
  - Network and USB cards facilitate **collaborative research** and **cloud-based analysis**

#A5 Ports and Connectors

---

## 1. Definition
- A **port** is a point where a **peripheral connects** to a system unit, allowing it to **send or receive data**.  
- Sometimes, **jack** is used to refer to audio or video ports.  
- Ports are located on the **front, back, or sides** of a computer, depending on the device type (desktop, notebook, Tablet PC).  
- A **connector** joins a cable to a port:  
  - One end connects to the **system unit**  
  - The other end connects to the **peripheral**

---


image.png

## 2. Common Ports

### USB Ports (Universal Serial Bus)
- Can connect **up to 127 devices** to a single connector  
- Typical devices: mouse, printer, digital camera, scanner, speakers, media players, smartphones, external drives  

**USB Connector Types:**

| Connector | Use |
|-----------|-----|
| Type A    | Desktop PCs, traditional notebooks, netbooks, Tablet PCs |
| Type B    | Peripherals (printers, scanners, external hard disks) |
| Mini-B    | Mobile devices (phones, cameras, handheld consoles) |

**USB Versions:**  
- USB 2.0 (Hi-Speed) → 40× faster than original USB  
- USB 3.0 → even faster, backward compatible  
- **USB hubs** allow multiple devices to connect to one port

### FireWire Ports (IEEE 1394)
- Supports multiple **high-speed devices**: cameras, VCRs, scanners, printers  
- Up to **63 devices per port**  
- Versions: FireWire 400, 800, 1600, 3200 (increasing speed)  
- **FireWire hubs** allow connecting multiple devices to one port

---

### Other Ports
| Port Type | Function |
|-----------|---------|

  - Power connection  
  - Slots for memory cards, optical drives, or other devices

#A6 Buses

---

## 1. Introduction
- Computers process and store data as **electronic bits**.  
- Bits move inside the computer along **electrical channels called buses**.  
- Analogy: A **bus is like a highway for data**, with bits traveling between components like vehicles moving between destinations.

---

## 2. Functions of a Bus
- Buses transfer bits between:  
  - **Input devices → Memory**  
  - **Memory → Processor**  
  - **Processor → Memory**  
  - **Memory → Output or Storage devices**  

- Buses consist of two main parts:  
  - **Data Bus:** Transfers the actual data  
  - **Address Bus:** Transfers information about where data should reside in memory

---

## 3. Bus Width
- Determines how many **bits can be transmitted at one time**  

**Examples:**  
- 32-bit bus → transmits 32 bits (4 bytes) at once  
- 64-bit bus → transmits 64 bits (8 bytes) at once  

> Wider buses allow **faster data transfer**, improving overall computer speed.  
> Most modern personal computers use a **64-bit bus**.

---

## 4. Bus Clock Speed
- Like processors, buses operate according to a **clock speed** measured in hertz (Hz)  
- **1 MHz = 1 million ticks per second**  

**Typical bus speeds:** 400, 533, 667, 800, 1066, 1333, 1600 MHz  

> Higher bus clock speed → faster data transmission → faster program execution

---

## 5. Types of Buses

| Bus Type | Function |
|----------|---------|
| System Bus (Front Side Bus – FSB) | Connects processor to main memory; often referred to simply as "the bus" |
| Backside Bus (BSB) | Connects processor to cache memory, improving speed for frequently used data |
| Expansion Bus | Connects processor to peripherals via adapter cards in expansion slots; allows devices like printers, external drives, or network cards to communicate efficiently with the CPU |

---

## 6. Practical Context
- **Financial Engineers:**  
  - High-frequency trading and **real-time risk calculations** rely on **wide, high-speed buses** to minimize latency when transferring large datasets between memory and processor  

- **Biostatisticians:**  
  - Analyzing **genomic, clinical, or epidemiological datasets** involves frequent memory-to-processor transfers  
  - **Bus width and speed** are critical for efficient computation

#A8 Power Supply in the System Unit

---

## 1. Introduction
- The **Power Supply Unit (PSU)** is a critical component of the system unit, responsible for **providing stable, usable electrical power** to all internal components.  
- Wall outlets deliver **alternating current (AC)** at 115–120 volts, whereas most computer components require **direct current (DC)** at 5–15 volts.  
- The PSU converts AC into the **precise DC voltages** required by the motherboard, processor, memory, storage devices, and peripherals.

---

## 2. Key Functions and Features

### Voltage Conversion
- Ensures **safe and stable voltage** for sensitive electronic components  
- Prevents **damage or operational instability** caused by voltage fluctuations or spikes

### Cooling Mechanisms
- Built-in **fans** dissipate heat generated during power conversion  
- Some fans have **variable speeds**  
- Efficient cooling improves the **reliability and lifespan** of both the PSU and the system

### Peripheral Power Management
- External devices (e.g., cable modems, printers, speakers) often use **AC adapters**  
- Modular power approach allows peripherals to draw the correct voltage independently  
- Prevents **overloading the system unit**

### Energy Efficiency and Safety
- Features include:  
  - Overcurrent protection  
  - Short-circuit protection  
  - Energy-efficient standards (e.g., **80 PLUS certification**)  
- Essential in **research environments** where computations may run continuously for hours or days

---

## 3. Conceptual Insight for Research Applications
- The PSU underpins **system stability and computational integrity**  
- High-performance tasks, such as:  
  - **Parallel simulations** in financial engineering  
  - **Processing large genomic datasets** in biostatistics  
  place **variable and high demands** on the CPU, GPU, and memory  
- A **robust PSU** ensures power fluctuations do not compromise **computational accuracy** or **system uptime**

> In essence, the PSU is the **foundation of the system’s operational reliability**: all other components depend on its ability to deliver **consistent and correctly regulated power**.


#A9 Putting It All Together: Components of the System Unit

---

## 1. Determinants of Computer Performance
The performance of a computer is governed by the interplay of several **key components** within the system unit:

- **Processor (CPU) Type and Architecture**  
  - Modern CPUs integrate **multiple cores** (dual-core, quad-core, or higher)  
  - Multi-core processors execute instructions concurrently, improving **throughput** and **energy efficiency**  

- **Clock Speed**  
  - The **system clock** regulates timing of all CPU operations  
  - Higher clock speeds allow more instructions per second, but overall performance also depends on **memory access times**, **bus width**, and **peripheral efficiency**  

- **Random Access Memory (RAM)**  
  - Volatile memory serves as the **workspace for active instructions and data**  
  - Sufficient RAM reduces reliance on slower storage, decreasing **latency** in execution  

- **Buses and Data Pathways**  
  - Bus **width** and **clock speed** influence the speed and volume of data transfer between CPU, memory, and peripherals  
  - Wider buses and higher frequencies improve **throughput**  

- **Storage Devices and Bays**  
  - Internal/external bays house drives for **persistent data storage**  
  - Fast storage (e.g., SSDs vs HDDs) reduces bottlenecks when reading/writing large datasets  

> **Contextual Insight:** For computationally intensive fields like **financial engineering** or **biostatistics**, optimizing CPU, memory, bus, and storage performance is critical for **real-time analytics**, **large-scale simulations**, and **complex modeling**.
