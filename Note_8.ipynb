## I. Overview of Networks

Networks are essential for **data collection, storage, and analysis** in biostatistics and financial engineering. A network is a collection of computers and devices connected through communication devices and transmission media, allowing sharing of:

- **Data & Information:** Patient records, real-time market feeds, simulation outputs  
- **Hardware:** Servers, printers, storage arrays  
- **Software Tools:** R, Python notebooks, MATLAB, or risk engines  
- **Funds Transfer & Transactions:** Critical in fintech applications  

Networks can be **internal** (hospital LAN, bank LAN) or **global** (Internet). Some organizations use **Value-Added Networks (VANs)** to ensure secure, reliable data exchange for sensitive clinical or financial information.

---

## II. Network Types and Scales

### 1. Local Area Network (LAN)
- Connects computers in a limited area (lab, office, trading floor)  
- **Nodes:** Devices on the network  
- **Shared Resources:** Printers, storage, analytical software  
- **Wireless LAN (WLAN):** Wireless extension of a wired LAN  

**Example:** KEMRI researchers use a WLAN to access a central clinical trial database on a wired LAN for real-time data entry and monitoring.

- Wearable or implanted sensors transmit patient or biometric data

- Financial Engineering: Biometric authentication for secure transactions

### 2. Metropolitan Area Network (MAN)
- Connects LANs within a city or region  
- Managed by consortia, ISPs, or governments  
- Supports city-wide disease surveillance or multi-branch financial processing

### 3. Wide Area Network (WAN)
- Spans large geographic areas, combining LANs and MANs  
- **Applications:**  
  - Biostatistics: Connect hospitals, labs, and global trial centers  
  - Financial Engineering: Connect trading floors, banks, and exchanges for real-time operations

#C.Communications over the Telephone Network

The **Public Switched Telephone Network (PSTN)** continues to support both voice and data communications globally. While historically for voice calls, modern digital upgrades allow PSTN to transmit critical data, from clinical trial datasets to real-time trading signals. Most of the infrastructure is digital, except the last-mile connection to end users, which may remain analog.

---

## I. Dial-Up Lines

- **Definition:** Temporary, analog-based connections, active only when needed  
- **Cost:** Comparable to a standard phone call  
- **Applications:**  
  - **Biostatistics:** Access remote health databases or rural clinical trial sites with limited connectivity  
  - **Financial Engineering:** Low-frequency data retrieval from branch offices in underdeveloped infrastructure  
- **Limitation:** Low transfer rates (~56 kbps), unsuitable for large genomic datasets or real-time market feeds

---

## II. Dedicated Lines

- **Definition:** Permanent, always-on connections between two endpoints  
- **Advantages:** Higher reliability, speed, and security than dial-up  
- **Types of Digital Dedicated Lines:**

| Line Type | Description | Domain Applications |
|-----------|------------|-------------------|
| **ISDN (Integrated Services Digital Network)** | Standard copper lines transmitting digital data | Biostatistics: Small clinics sending lab results; Finance: Remote offices uploading reports |
| **DSL (Digital Subscriber Line)** | High-speed over copper; ADSL favors downloads | Biostatistics: Accessing public datasets; Finance: Receiving market feeds |
| **FTTP (Fiber to the Premises)** | Fiber-optic directly to building/home; includes FTTH/FTTB | Biostatistics: High-resolution imaging/genomic data transfer; Finance: Low-latency trading infrastructure |
| **T-Carrier Lines (T1, T3)** | High-speed long-distance digital lines carrying multiple signals | Biostatistics: Multi-site data aggregation; Finance: Trading hubs & exchange connections |
| **ATM (Asynchronous Transfer Mode)** | Ultra-high-speed service carrying voice, video, data | Biostatistics: Telemedicine and streaming clinical data; Finance: Real-time algorithmic trading |

---

## III. Comparison and Cost Considerations

| Feature | Dial-Up | DSL / ISDN | FTTP | T1/T3 | ATM |
|---------|--------|------------|------|-------|-----|
| **Connection** | Temporary | Permanent | Permanent | Permanent | Permanent |
| **Speed** | Low (~56 kbps) | Moderate (256 kbps–10 Mbps) | High (100 Mbps–1 Gbps+) | Very High (1.544–44.736 Mbps) | Ultra-High |
| **Cost** | Low | Moderate | High | High | Very High |
| **Typical Users** | Small clinics, remote offices | Small businesses, labs | Hospitals, offices | Large enterprises, trading floors | Large networks, critical services |
| **Domain Applications** | Remote data retrieval | Clinical trials, moderate market data | Genomics, telemedicine, algorithmic trading | Financial exchanges, research hubs | Multimedia, high-volume streaming |

**Insight:**  
- Biostatistics: Speed and data integrity are critical for multi-terabyte genomic or imaging datasets  
- Financial Engineering: Low latency and uptime are paramount; fiber or T-carrier lines are essential

---

## IV. Integration with Other Internet Connections

- **Cable Television (CATV) Lines** and **Fixed Wireless Systems** provide alternative high-speed access  
- **Fixed Wireless:** Antennas transmit signals between buildings/towers without cabling  
- Complements PSTN, especially in rural research labs or distributed financial offices

#F Communications Channels

A **communications channel** is the pathway through which data, instructions, or information travel between devices. For financial engineers and biostatisticians, these channels are more than “pipes”; they directly influence **speed, accuracy, and feasibility** of computations, especially for large or time-sensitive datasets.

---

## 1. Transmission Media and Their Importance

Transmission media are the **physical or wireless materials** that carry signals. The choice of media determines how fast and reliably data moves—whether it’s **market tick data** for algorithmic trading or **multi-terabyte epidemiological datasets** for health modeling.

| Type        | Description                          | Financial Engineering Relevance                                   | Biostatistics / Health Data Relevance                          |
|------------|--------------------------------------|------------------------------------------------------------------|----------------------------------------------------------------|
| Physical (wired) | Copper wires, fiber optics, coaxial cables | Fiber-optic or T1/T3 lines enable low-latency connections to exchanges or trading APIs; essential for high-frequency trading (HFT) | Fiber or dedicated lines allow rapid transfer of large genomic datasets or longitudinal survey data |
| Wireless   | Radio waves, microwaves, infrared, satellite signals | Mobile trading platforms, remote terminals                        | Field data collection, remote research stations, health monitoring devices |

---

## 2. Bandwidth Considerations

- **Financial Engineering:**  
  High-frequency trading and real-time market analytics require **high bandwidth and low-latency channels**. Even millisecond delays can impact trades, pricing, or risk hedging.

- **Biostatistics:**  
  Large-scale simulations, climate-health models, or genomic pipelines demand **high-throughput channels**, as datasets may reach terabytes. Low-bandwidth channels slow computation and extend project timelines.  

**Example:** Downloading a multi-year malaria surveillance dataset from KEMRI servers over a 1 Mbps DSL line may take hours, while a 100 Mbps fiber connection reduces this to minutes.

---

## 3. How Data Travels Across Channels

1. **Request Initiation:** A financial application or R/Python script sends a data request over physical media (fiber, copper) or wireless media (4G, Wi-Fi).  
2. **ISP to Backbone:** The request passes through local ISP lines, T1 connections, microwave links, or satellite uplinks to reach the Internet backbone.  
3. **Backbone Transit:** High-capacity T3 lines or optical fiber carry data between major servers and data centers.  
4. **Destination Server:** Data arrives at the target server, such as a market data provider or a public health database.  

**Insight:** Each “hop” introduces potential **latency or bandwidth constraints**. Financial engineers prioritize **low-latency hops**, whereas biostatisticians prioritize **high-throughput paths**.

---

## 4. Broadband vs. Narrowband

- **Broadband:** Transmits multiple signals simultaneously, enabling **faster, parallelized data flows**.  
  - Financial engineers use fiber or leased lines to process multiple market feeds in real time.  
  - Biostatisticians use broadband for rapid transfer of genomic or survey datasets.  

- **Narrowband:** Dial-up or basic DSL, suitable only for **small datasets** or **non-time-sensitive transactions**.

# G. Physical Transmission Media  
Physical transmission media are the **tangible infrastructures** through which digital signals travel in a communications system. In **financial engineering** and **biostatistics**, these media underpin data-intensive computation, real-time analytics, and distributed research collaboration. Decisions made at this physical layer ultimately shape **latency**, **throughput**, **reliability**, and **data integrity**—all critical to quantitative work.

The three principal physical transmission media are **twisted-pair cable**, **coaxial cable**, and **fiber-optic cable**. These are commonly deployed within buildings, across campuses, or underground between facilities, and they form the backbone of **Ethernet-based local area networks (LANs)**.

---

## 1. Twisted-Pair Cable

Twisted-pair cable consists of pairs of insulated copper wires twisted together to reduce **electromagnetic interference (noise)**. Noise introduces signal distortion, which may cause data corruption, retransmissions, or latency spikes.

### Applied Context

**Financial Engineering**  
Twisted-pair cabling is widely used within offices, trading floors, and fintech startups where Ethernet LANs connect workstations, servers, and local switches. While adequate for routine analytics, it is not optimal for **ultra-low-latency trading environments**, where even microsecond-level delays can affect execution quality.

**Biostatistics**  
In research laboratories and teaching hospitals, twisted-pair cables support internal networks used for statistical analysis, electronic health records, and collaborative modeling. For routine regression analysis or dashboarding, twisted-pair is sufficient; however, **large-scale simulations or genomic pipelines** may expose its bandwidth limitations.

### Key Insight

Twisted-pair cable is a **cost-effective but performance-limited** medium—appropriate for standard workloads but unsuitable for extreme data volumes or latency-sensitive tasks.

---

## 2. Coaxial Cable

Coaxial cable uses a single copper conductor surrounded by insulating and shielding layers that reduce signal leakage and interference. Historically, it supported longer distances than twisted-pair cable and remains common in **cable television (CATV)** infrastructure.

### Applied Context

**Financial Engineering**  
Coaxial cable is rarely used in modern professional financial networks. However, **cable Internet connections**—common among retail traders or home-based quantitative analysts—often rely on coaxial infrastructure. These connections may experience **shared-bandwidth congestion**, leading to unpredictable latency during peak usage.

**Biostatistics**  
Home-based researchers, graduate students, or field analysts often rely on cable broadband delivered over coaxial lines. While adequate for downloading datasets or submitting analyses, performance can degrade when many users share the same neighborhood infrastructure.

### Key Insight

Coaxial cable represents a **transitional technology**—functional but increasingly outperformed by fiber-optic alternatives in professional research and financial environments.

---

## 3. Fiber-Optic Cable

Fiber-optic cable transmits data as **pulses of light** through thin strands of glass or plastic (optical fibers). This medium supports **extremely high bandwidth**, **minimal signal loss**, and strong resistance to electromagnetic interference.

### Advantages in Quantitative Disciplines

**Financial Engineering**  
Fiber optics are foundational to:
- High-frequency trading (HFT)  
- Real-time market data feeds  
- Low-latency links between exchanges and data centers  

In global finance, fiber routes are strategically optimized because **physical distance and signal speed directly influence competitive advantage**.

**Biostatistics**  
Fiber infrastructure enables:
- Transfer of multi-terabyte genomic datasets  
- Distributed Bayesian simulations  
- Cloud-based epidemiological modeling  

This is especially important for collaboration across institutions and integration with national or international surveillance systems.

### Kenya / Nairobi Context

Fiber-optic backbones now connect:
- Financial institutions around the **Nairobi Securities Exchange (NSE)**  
- Universities and research institutes (e.g., **KEMRI**, public universities)  
- International data centers supporting cloud analytics  

This expansion has significantly reduced the digital divide for advanced analytics, although installation costs still limit universal access.

### Limitations

- Higher installation and maintenance costs  
- Greater technical complexity in setup and modification  

Despite these challenges, fiber is rapidly replacing legacy copper infrastructure due to its **future-proof capacity**.

## Conceptual Takeaway for Practitioners

For **financial engineers** and **biostatisticians**, physical transmission media are not neutral technical details. They determine:

- How fast models can ingest data  
- Whether real-time decisions are feasible  
- How reliably sensitive health or financial data move across systems  

In advanced quantitative work, **fiber-optic infrastructure** is increasingly a **methodological enabler**, not merely an IT choice.

# H. Wireless Transmission Media  

Wireless transmission media enable data communication **without physical cables**, using **electromagnetic waves** to transmit signals through air or space. In contemporary quantitative practice, wireless infrastructure supports **mobility**, **real-time decision systems**, **distributed analytics**, **telemedicine**, **mobile trading platforms**, and **field-based data collection**. The trade-off between convenience and performance—particularly **latency**, **reliability**, and **security**—is central to evaluating wireless technologies.

Wireless transmission media include **infrared**, **broadcast radio**, **cellular radio**, **microwaves**, and **communications satellites**.

---

## 1. Infrared (IR) Transmission

Infrared transmission uses light waves below the visible spectrum to transmit data over short distances. It typically requires **direct line-of-sight** and is used for device-to-device communication.

### Applied Context

**Financial Engineering**  
Infrared is rarely used in trading or analytics systems but may appear in peripheral device communication (e.g., secure authentication hardware or point-of-sale systems).

**Biostatistics**  
IR supports local data transfer between laboratory devices (e.g., sensors, printers, handheld diagnostic tools). Its short range limits use in large research networks.

### Key Insight

Infrared is suited for **localized, low-power, short-range data transfer** and is not scalable for analytics infrastructures.

---

## 2. Broadcast Radio Transmission

Broadcast radio distributes signals over short or long distances and underpins technologies such as **Wi-Fi**, **Bluetooth**, **Ultra-Wideband (UWB)**, and **WiMAX**.

### Applied Context

**Financial Engineering**  
Wi-Fi enables:
- Mobile trading platforms  
- Remote access to financial dashboards  
- Cloud-based portfolio analytics  

However, Wi-Fi introduces **variable latency** and **security risks**, making it unsuitable for **high-frequency trading (HFT)** systems.

**Biostatistics**  
Wi-Fi supports:
- Hospital information systems  
- Research collaboration platforms  
- Real-time epidemiological dashboards  

Bluetooth enables wearable health devices that feed continuous patient data into statistical models.

### Kenya / Nairobi Context

Universities, fintech hubs, and hospitals in Nairobi increasingly rely on Wi-Fi-enabled networks for:
- Digital health surveillance (e.g., DHIS2 reporting)  
- Mobile banking analytics (e.g., M-Pesa ecosystem modeling)  
- Remote learning and teleconsultation  

### Key Insight

Broadcast radio enables **ubiquitous connectivity**, but performance fluctuates due to congestion and interference.

---

## 3. Cellular Radio Transmission

Cellular radio uses high-frequency radio waves to connect mobile devices to distributed base stations, enabling voice and data transmission over wide geographic areas (e.g., **3G**, **4G**, **5G**).

### Applied Context

**Financial Engineering**  
Cellular networks enable:
- Mobile trading applications  
- Real-time payment platforms  
- Market data streaming  

In Kenya, mobile financial systems dominate economic activity, generating massive transaction datasets for **risk modeling**, **fraud detection**, and **liquidity forecasting**.

**Biostatistics**  
Cellular connectivity supports:
- Mobile health (mHealth) reporting  
- Disease surveillance via SMS platforms  
- Remote clinical trial monitoring  

Field epidemiologists transmit data from rural areas directly to centralized analytic platforms.

### Key Insight

Cellular transmission is critical for **distributed data ecosystems**, especially in low-infrastructure environments.

---

## 4. Microwave Transmission (Fixed Wireless)

Microwaves transmit high-frequency radio signals via **line-of-sight (LOS)** links between towers or stations. They deliver high-speed communication where cabling is impractical.

### Applied Context

**Financial Engineering**  
Microwave links are used in **low-latency trading networks** between financial centers. Globally, firms deploy microwave routes to reduce signal propagation delays.

**Biostatistics**  
Microwave networks connect:
- University campuses  
- Hospitals  
- Government data centers  

supporting large dataset transfers and telemedicine systems.

### Kenya / Regional Context

Microwave links connect:
- Remote counties to national health data systems  
- Rural banks to core financial networks  
- Universities and research institutes to national backbones  

### Key Insight

Microwave transmission balances **speed** and **infrastructure feasibility**, especially in geographically dispersed regions.

---

## 5. Communications Satellites

Communications satellites relay microwave signals between earth stations, enabling **global connectivity**. Transmission involves:
- **Uplink:** Earth → Satellite  
- **Downlink:** Satellite → Earth  

### Applied Context

**Financial Engineering**  
Satellites support:
- Global financial communications  
- Maritime and aviation market data  
- International payment systems  

Latency remains higher than fiber, limiting use in time-critical trading.

**Biostatistics**  
Satellites enable:
- Remote health surveillance  
- Disaster response data transmission  
- Climate–health monitoring  
- Telemedicine in rural regions  

### Kenya Context

Satellite Internet connects:
- Remote health facilities in arid counties  
- Wildlife and environmental monitoring stations  
- Humanitarian field operations  

### Key Insight

Satellites ensure **connectivity equity**, though cost and latency constrain performance.

---

## Ethics & Public Health Considerations: Wireless Radiation

Public concern exists regarding potential health risks from:
- Cell phones  
- Cellular towers  
- Wi-Fi devices  

### Scientific Evidence

- No conclusive causal link has been established between typical exposure levels and adverse health outcomes.  
- Long-term epidemiological studies are ongoing.  
- Regulatory exposure limits are based on established safety thresholds.  

### Policy Implications

Governments balance:
- Public health concerns  
- Digital inclusion  
- Economic growth  

In many jurisdictions, infrastructure rollout is prioritized to support **financial inclusion** and **health system digitization**.

---

## Analytical Perspective

**For Biostatisticians**  
This debate highlights:
- Challenges of causal inference in low-exposure epidemiology  
- Need for longitudinal cohort studies  
- Risk communication under uncertainty  

**For Financial Engineers**  
Wireless infrastructure drives:
- Digital financial inclusion  
- Market efficiency  
- Platform-based economic growth  

Delays in deployment can impose significant **macroeconomic opportunity costs**.
