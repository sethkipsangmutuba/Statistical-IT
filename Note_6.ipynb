##B2 Hard Disks (HDDs)

### Structure and Usage

- Composed of **rigid circular platters** coated with magnetic material  
- **Internal HDDs:** Fixed in servers or desktops  
- **External HDDs:** Portable for dataset transfer between workstations  
- **Capacities:** Personal systems (160 GB–2 TB), enterprise servers (tens of TBs)

### Applications

- **Biostatisticians:** Store raw and processed patient/experimental data, code repositories, intermediate simulation outputs  
- **Financial engineers:** Maintain transaction logs, historical datasets, portfolio snapshots, and model parameters

### Data Recording

- **Longitudinal (horizontal particle alignment):** Traditional HDDs  
- **Perpendicular (vertical particle alignment):** Higher density (~10× increase)  

**Use Cases:**  
- Omics matrices in biostatistics  
- Tick-level market data in finance

### Read/Write Operations

- **Writing:** Data transfer from memory → storage  
- **Reading:** Data transfer from storage → memory  

HDDs serve as both **input** and **output** sources but are classified as **storage devices**.

### Access Time

- Slower than memory: **milliseconds (ms)** or **microseconds (µs)**  
- In computational pipelines or financial backtesting, access speed affects simulation runtime and model responsiveness

# Characteristics of a Hard Disk


---

## Overview

A **hard disk (HDD)** is a fundamental storage device, essential for:

- Storing **large datasets**  
- Maintaining **financial transaction logs**  
- Preserving **statistical and financial models**

Understanding HDD characteristics is critical for:

- Biostatisticians handling **genomic data** or **clinical trials**  
- Financial engineers managing **high-frequency trading datasets** or **risk simulations**

---

## Key Characteristics of Hard Disks

### 1. Capacity

- Determined by:  
  - **Magnetic recording type** (longitudinal vs. perpendicular)  
  - **Number of platters**  
  - **Magnetic coating**
- Platters: Aluminum, glass, or ceramic, coated with a magnetic alloy (~3 millionths of an inch thick)  
- **Example:** 1 TB HDD can store approximately:  
  - 500 million pages of text (clinical notes)  
  - 285,000 digital photos (medical imaging)  
  - 250,000 audio files (patient interviews)  
  - 120 hours of video (training simulations)

---

### 2. Platters, Tracks, and Sectors

- **Track:** Circular recording band on a platter  
- **Sector:** Pie-shaped segment of a track; typically stores 512 bytes  
- Multiple platters are stacked, each with read/write heads on both sides  
- **Cylinder:** Vertical alignment of tracks across platters; accessing one cylinder retrieves all corresponding tracks simultaneously  

**Relevance:** Cylinder-based access accelerates retrieval of genomic sequences or batch financial transactions

---

### 3. Read/Write Heads

- Float above platters on an air cushion (~2 millionths of an inch)  
- **Precision is critical**: Dust, hair, or smoke can cause head crashes → data loss  
- **Domain note:** In clinical or financial databases, head crashes compromise research integrity or auditing; **regular backups are mandatory**

---

### 4. Rotational Speed

- Platters spin at **5,400–15,000 RPM**  
- Higher RPM reduces access time for near-instant data retrieval  

**Finance application:** High RPM disks support rapid historical trading data access for real-time risk assessments

---

### 5. Access Time and Transfer Rate

- **Access time:** 3–12 ms (modern disks)  
- **Transfer rate:** Example 300 MBps  

**Biostatistics application:** Faster rates enable quick loading of multi-terabyte sequencing datasets

---

### 6. RAID (Redundant Array of Independent Disks)

- Combines multiple disks into a **single logical unit**  
- Provides **speed**, **redundancy**, and **reliability**  
- Critical for: Clinical trial data storage, back-testing financial models, continuous access to large datasets

---

### 7. NAS (Network-Attached Storage)

- Centralized storage accessed via a **network**, often with RAID  
- Small NAS devices (6 TB+) are accessible to individual labs or small financial teams

---

### 8. External and Removable Hard Disks

- **External disks:** USB/FireWire, sealed cases, typically up to 4 TB  
- **Removable disks:** Inserted/removed as needed, typically up to 1 TB  
- **Use cases:** Transport patient datasets, archive trading histories, secure financial models offline

---

### 9. Miniature Hard Disks

- Form factors: 1.8 inch, 1 inch, 0.85 inch  
- Found in **portable devices**, media players, smartphones  
- Capacity: 4–250 GB  
- Useful for **mobile data collection** or **field studies**

---

### 10. Hard Disk Controllers and Interfaces

- Control **data transfer** between disk and system bus  
- Common Interfaces: SATA, eSATA, EIDE, SCSI, SAS  
  - **SATA:** Serial, flexible cables, common in PCs  
  - **EIDE:** Parallel, supports multiple disks  
  - **SCSI/SAS:** Enterprise-grade, high throughput, reliable  

**Domain relevance:** SAS arrays used for **high-throughput biostatistical simulations** or storing **transactional logs in hedge funds**

---

### 11. Maintenance

- Typical lifespan: 3–5 years  
- Preventive measures: Defragmentation, error scanning, and regular backups  
- Crucial for protecting: **clinical research data**, **patient records**, **financial audit trails**

---

## Key Takeaways

- Disk performance directly affects **data processing**, **model simulation speed**, and **database reliability**  
- **RAID/NAS configurations** enhance reliability and accessibility for mission-critical datasets  
- **Backups and preventive maintenance** are non-negotiable to avoid catastrophic data loss  
- **Interface choice (SATA vs SAS)** significantly impacts **transfer rates** in high-volume environments


#B3 Flash Memory Storage

## Overview

**Flash memory** is a type of **nonvolatile storage**, retaining data even when power is off.  
Unlike magnetic hard disks or optical discs, flash memory:

- Has **no moving parts** → more durable, shock-resistant, energy-efficient  
- Ideal for **frequently accessed datasets** and models in biostatistics and finance  

**Use Cases:**

- Biostatisticians: Longitudinal patient datasets, genomic sequences  
- Financial engineers: Continuous high-frequency trading simulations, risk models

---

## Types of Flash Memory Storage

### 1. Solid-State Drives (SSDs)

- High-speed storage using flash memory  
- **Form factors:** 3.5-inch, 2.5-inch, 1.8-inch; servers, desktops, laptops, mobile devices  
- **Capacities:** 16 GB – 256 GB+  

**Advantages:**

- Ultra-fast access (~0.1 ms, ~80× faster than HDDs)  
- High transfer rates → accelerate genomic or financial dataset loading  
- Energy-efficient → less heat, lower power consumption, ideal for 24/7 simulations  
- Long lifespan → up to 50 years vs. 3–5 years for HDDs  

**Disadvantages:**

- Higher cost per GB  
- Slightly higher failure rate than HDDs (cost decreasing over time)

---

### 2. Memory Cards

- Small, removable flash storage devices  
- **Common types:** CompactFlash (CF), SD, SDHC, microSD, microSDHC, xD Picture Card, Memory Stick, M2  
- **Capacity/Speed:** Varies; suitable for **mobile research devices**, point-of-care medical equipment, or portable market data storage  

**Domain Example:**

- Biostatistics: Field surveys stored on microSD cards  
- Finance: Carry intraday trading logs between offices

---

### 3. USB Flash Drives

- Compact, portable, and durable  
- Connect via USB → ideal for **quick data transfer**, **emergency backups**, **secure movement of sensitive datasets**  
- Resistant to airport X-ray screening; handle carefully to avoid physical damage  

---

### 4. ExpressCard Modules

- Removable devices (~75 mm × 34 mm; L-shaped 54 mm wide) for notebook ExpressCard slots  
- Extend laptop functionality: storage expansion, security, multimedia interfaces  
- Useful for **mobile labs** in biostatistics or **notebook-based trading setups** in finance  

---

## Domain Applications

| Field | Application |
|-------|------------|
| Biostatistics | SSDs accelerate genomic sequence analysis, clinical trial processing, and mobile data collection. Memory cards store and transport field datasets securely. |
| Financial Engineering | High-speed SSDs reduce latency for retrieving historical market data and real-time risk modeling. ExpressCard modules allow secure notebook-based trading setups. |

---

## Key Takeaways

- Flash memory is **fast**, **durable**, and **energy-efficient** → ideal for high-intensity computational environments  
- **Type selection** (SSD, memory card, USB, ExpressCard) depends on:  
  - Capacity needs  
  - Speed requirements  
  - Portability constraints  
- **Regular backups** remain critical to prevent data loss, even with nonvolatile flash storage


#B4 Optical Discs

---

## Overview

**Optical discs** are flat, circular storage media made of metal, plastic, and lacquer:

- Diameter: ~4.75 inches  
- Thickness: <1/20th inch  

**Key features:**

- Portable, durable, cost-effective  
- Store software, data, digital photos, audio, video  
- Use **laser technology** to read/write data via microscopic **pits** and **lands**  

**Domain Applications:**

- **Biostatistics:** Archive datasets, statistical software, longitudinal study results  
- **Financial Engineering:** Store historical market datasets, simulations, or software snapshots  

> Slower than SSDs/flash memory but suitable for **long-term, low-cost archival storage**

---

## Optical Disc Types and Characteristics

### 1. Read-Only Discs (ROM)

- Examples: **CD-ROM, DVD-ROM, BD-ROM**  
- Data can only be **read**, not written/erased  
- Uses: Preloaded software, databases, reference datasets  
- Capacities:  
  - CD-ROM: 650 MB–1 GB  
  - DVD-ROM: 4.7–17 GB  
  - Blu-ray (BD-ROM): 25–100+ GB  

---

### 2. Recordable Discs (R)

- Examples: **CD-R, DVD-R, BD-R**  
- Data can be **written once**, not erased  
- Uses: Archiving finalized datasets, models, or reports  

---

### 3. Rewritable Discs (RW/RE)

- Examples: **CD-RW, DVD-RW, DVD+RW, BD-RE**  
- Data can be **written and erased multiple times**  
- Uses: Temporary storage during iterative analyses or frequent model updates  

---

### 4. Single-Sided vs Double-Sided

- **Single-sided:** Data on one side; label on opposite side  
- **Double-sided:** Data on both sides → maximizes archival capacity without increasing storage space  

---

## How Optical Discs Work

- **Data Storage Mechanism:**  
  - High-powered laser creates **pits** in the disc  
  - Lower-powered laser reads data by reflecting off **pits** and **lands**  
  - Reflected light converted into **digital bits** for computer processing  

- **Durability:** 5–100 years with proper care  
  - Avoid touching underside, heat, sunlight, stacking, or eating/drinking near discs  

- **Labeling Technology:**  
  - Traditional: Adhesive or silk-screened labels  
  - LightScribe: Etches labels directly onto specially coated discs  

---

## Specialized Optical Disc Uses

- **Archive Discs & Picture CDs:** Store large collections of digital images or scanned medical forms; useful for dataset snapshots  
- **High-Capacity Discs:**  
  - DVDs: Higher density than CDs; support double layers and double-sided formats  
  - Blu-ray (BD): 100 GB+, ideal for high-resolution medical images, genomic datasets, or complex simulations  
- **Mini Discs (UMD, mini-BD):** Portable, used in handheld devices or data loggers  

---

## Practical Notes for Biostatisticians & Financial Engineers

- **Archival Storage:** Best for long-term, infrequently accessed data (e.g., historical market records, old clinical trial datasets)  
- **Backup Strategy:** Optical discs are scratch-prone → maintain redundant backups on SSDs, flash drives, or cloud  
- **Data Portability:** Useful for offline transfer between labs, financial offices, or secure research facilities  
- **Data Security:** Physical control (secure cabinets/safes) provides access protection  

---

## Key Takeaways

- Optical discs are **durable** and **long-term**, but slower than SSDs or flash memory  
- **Use ROMs** for distribution, **R discs** for finalized archival, **RW/RE** for iterative projects  
- Proper handling and storage **extend lifespan** and preserve data integrity  
- For **high-volume datasets**, Blu-ray or multi-layer DVDs provide much greater capacity than standard CDs

# Other Types of Storage

---

While **hard disks**, **flash memory**, and **optical discs** handle most personal and enterprise storage needs, **specialized storage options** exist for archival, secure, and large-scale applications. These include **tape storage**, **magnetic stripe and smart cards**, **microfilm/microfiche**, and **enterprise storage systems**.

---

## 1. Tape Storage

**Description:**  
Tape storage uses a magnetically coated ribbon of plastic. It originated in early mainframe systems.

**Form:**

- Modern systems: tape cartridges (small rectangular cases)
- Older systems: reel-to-reel tapes

**Use Case:**

- Long-term storage and backup
- Best suited for very large datasets with infrequent access

**Access Type:**

- **Sequential access** – data must be read in order
- Slower than direct-access media such as HDDs or SSDs

**Relevance for Biostatisticians:**

- Archiving clinical trial data
- Longitudinal patient datasets
- Raw genomic sequences

**Relevance for Financial Engineers:**

- Archiving historical market data
- Long-term storage of risk simulations
- Regulatory and audit reports

---

## 2. Magnetic Stripe Cards and Smart Cards

### Magnetic Stripe Cards

- Store identification data (e.g., name, account number, expiration date)
- Read using a magnetic stripe reader
- Common in:
  - Financial transactions
  - Physical access control systems

### Smart Cards

- Contain an embedded **microprocessor**
- Support storage, processing, and input/output
- Require a specialized reader

**Applications:**

- **Medical:** patient records, vaccination history
- **Financial:** prepaid transactions, secure banking, digital wallets
- **Security:** campus access, system authentication

**Ethical Considerations:**

- Smart cards generate traceable transaction records
- Raise privacy and data protection concerns
- Increasing integration with biometrics (e.g., fingerprints)

---

## 3. Enterprise Storage

**Definition:**  
High-capacity, high-availability storage systems designed for large organizations with hundreds or thousands of users.

**Key Features:**

- High redundancy and fault tolerance
- Dedicated storage networks:
  - **SANs (Storage Area Networks)**
- Capacity ranging into hundreds of terabytes (or more)
- Often integrates:
  - Optical disc servers
  - Tape libraries

**Relevance:**

- Essential for hospitals, banks, financial institutions, and global research labs
- Supports high-speed access to massive datasets

---

## 4. Microfilm and Microfiche

**Description:**  
Film-based storage that preserves documents as microscopic images.

- **Microfilm:** roll film (100–215 feet)
- **Microfiche:** flat sheets (~4 × 6 inches)

**Access:**

- Requires specialized microfilm or microfiche readers

**Common Uses:**

- Archiving newspapers
- Genealogical records
- Canceled checks
- Inactive corporate or government files

**Advantages:**

- Longest lifespan of any storage medium
- Extremely space-efficient
- Low-cost archival solution

**Relevance:**

- Archival research records
- Historical financial transactions
- Rare or irreplaceable clinical data

---

## 5. Long-Term Preservation Example: The Rosetta Project

- **Rosetta Disk:** a 3-inch nickel disk
- Preserves over **1,000 languages** etched microscopically
- Designed to last for thousands of years

**Relevance:**

- Demonstrates ultra-long-term data preservation
- Illustrates how biological, linguistic, or financial data could be stored for centuries without digital degradation

# Putting It All Together: Strategic Storage Selection

Selecting an appropriate **storage solution** depends on three interrelated factors that jointly determine performance, reliability, and sustainability in data-intensive environments such as **biostatistics** and **financial engineering**.

---

## Core Factors in Storage Selection

### 1. Volume of Data

The **amount of data, instructions, or multimedia** to be stored dictates the choice of storage:

- **Small-scale / portable needs:**  
  USB flash drives, memory cards  
- **Large-scale / institutional needs:**  
  NAS, SAN, tape libraries, enterprise HDD/SSD arrays  

As datasets grow from gigabytes to petabytes, storage must scale accordingly.

---

### 2. Hardware and Software Compatibility

Storage solutions must integrate seamlessly with existing systems:

- Operating systems and file systems  
- Network infrastructure  
- Interface standards:
  - SATA
  - SAS
  - USB
  - Thunderbolt
- RAID and redundancy compatibility  

Incompatible storage can bottleneck analytics pipelines or compromise reliability.

---

### 3. Cost and Longevity

Economic constraints must be balanced against performance and lifespan:

- **SSDs:**  
  High-speed access for real-time analytics, higher cost per terabyte  
- **HDDs:**  
  Cost-effective for bulk storage, slower access  
- **Microfilm / archival media:**  
  Extremely cost-effective for centuries-long preservation  

**Trade-off principle:** speed vs. cost vs. durability.

## Key Storage Management Practices

Effective storage depends on **maintenance**, not just hardware choice.

### 1. Disk Cleanup

- Removes temporary files and unused applications  
- Frees space and improves efficiency  

### 2. Disk Defragmentation

- Consolidates fragmented files into contiguous blocks  
- Reduces access time for HDDs  
- Less relevant for SSDs due to non-mechanical access  

### 3. Backup Strategies

- Routine backups using:
  - External drives
  - Tape
  - Cloud services  
- Use RAID, NAS, and SAN for redundancy and high availability  

### 4. Practical Troubleshooting

Common issues include:

- HDD head crashes  
- Missing or faulty cables  
- Corrupted memory cards  
- Unrecognized external drives  

**Proactive monitoring and preventive maintenance** are essential.
